```python
def get_seq(uniprot_id):
    url = f"http://www.uniprot.org/uniprot/{uniprot_id}.fasta"
    response = requests.post(url)
    cData=''.join(response.text)
    Seq=StringIO(cData)
    return MutableSeq(str(list(SeqIO.parse(Seq,'fasta'))[0].seq))
```

```python
## create a dictionary, dict_name_uniprot, where the keys are kinase names and the values are the corresponding Uniprot IDs, and save it to a pickle file
mg = mygene.MyGeneInfo()
results = mg.querymany(ls_symbol, scopes='symbol', fields='uniprot', species='human')

# no idea why there are two kinas TEC in results, so remove the invalid one
for i in range(len(results)):
    try:
        if results[i]['_id'] == '100124696':
            del results[i]
            break
    except:
        pass

# check if the number of results is equal to the number of kinases
assert len(results) == len(ls_name)

dict_name_uniprot = {}

for res, symbol_name in zip(results, ls_symbol_name):
    symbol, name = symbol_name
    assert res['query'] == symbol
    if 'notfound' in res:
        dict_name_uniprot[name] = None
    else:
        try:
            dict_name_uniprot[name] = res['uniprot']['Swiss-Prot']
        except:
            dict_name_uniprot[name] = None

# manually update the Uniprot IDs which is not found by mygene 
dict_updation = {'ADCK3': 'Q8NI60', 'ADCK4': 'Q96D53', 'CDC2L1': 'P21127', 'CDC2L2': 'Q9UQ88', 'CDC2L5': 'Q14004', 'CDK11': 'Q9BWU1', 'ICK': 'Q9UPZ9', 'MST4': 'Q499L9', 'MTOR': 'P42345', 'NIM1': 'Q8IY84', 'PAK7': 'Q9P286', 'PCTK1': 'Q00536', 'PCTK2': 'Q00537', 'PCTK3': 'Q07002', 'PFTK1': 'O94921', 'QSK': 'Q9Y2K2', 'RIPK4': 'P57078', 'RIPK5': 'Q6XUX3', 'SgK110': 'P0C264', 'YSK4': 'Q56UN5', 'PRP4': 'Q13523', 'PFCDPK1(Pfalciparum)': 'P62344', 'PFPK5(Pfalciparum)': 'P61075', 'PFTAIRE2': 'Q96Q40', 'PKNB(Mtuberculosis)': 'P9WI81', 'ZAK': 'Q9NYL2'}
for name in dict_name_uniprot:
    if (name in dict_updation) and dict_name_uniprot[name] == None:
        dict_name_uniprot[name] = dict_updation[name]

with open('dict_name_uniprot.pkl', 'wb') as f:
    pickle.dump(dict_name_uniprot, f)
```
